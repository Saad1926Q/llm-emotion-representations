# Emotion experiment constants

This notebook freezes the model identifier, target emotion taxonomy, and Dataset 1 generation templates before any prompts or activations are generated.

```python
# Primary model for the experiment.
MODEL_ID = "Qwen/Qwen3.5-2B"

print(MODEL_ID)
```

## Target emotions

The ten target labels preserve useful distinctions within negative and positive emotion. `neutral` is a control class, not a target emotion. The classifier-facing labels are kept explicit so later notebooks use the same taxonomy.

```python
TARGET_EMOTIONS = [
    "anger",
    "annoyance",
    "fear",
    "nervousness",
    "sadness",
    "disappointment",
    "joy",
    "excitement",
    "admiration",
    "gratitude",
]

NEUTRAL_LABEL = "neutral"
EMOTION_LABELS = TARGET_EMOTIONS + [NEUTRAL_LABEL]

assert len(TARGET_EMOTIONS) == 10
assert len(set(EMOTION_LABELS)) == len(EMOTION_LABELS)
print(EMOTION_LABELS)
```

## Dataset 1 templates

Each template creates one short independent story. `{emotion}` and `{domain}` are filled when the manifest is built. The requested emotion appears in the instruction so the model produces enough emotion-related material, but the story must not name that emotion.

```python
TEMPLATES = {
    "A": (
        "Write a short first-person story that conveys {emotion}. "
        "Show it through what happens rather than naming it. "
        "Set the story in {domain}. Write 80-120 words. "
        "Do not use the word {emotion} or discuss these instructions."
    ),
    "B": (
        "Describe a realistic personal experience involving {emotion}. "
        "Use concrete events and reactions without naming the emotion. "
        "Set the experience in {domain}. Write 80-120 words. "
        "Do not discuss these instructions."
    ),
    "C": (
        "Write about a situation in {domain} that would naturally cause {emotion}. "
        "Make the response implicit rather than directly explaining the feeling. "
        "Write 80-120 words. Do not use the word {emotion} or discuss these instructions."
    ),
    "D": (
        "Write a brief diary entry reflecting {emotion}. "
        "Express it through the description of the event and the narrator's actions. "
        "Set the diary entry in {domain}. Write 80-120 words. "
        "Do not use the word {emotion} or discuss these instructions."
    ),
    "E": (
        "Generate a natural short narrative in {domain} in which the central emotional state is {emotion}. "
        "Avoid explicitly labeling the emotional state. Write 80-120 words. "
        "Do not use the word {emotion} or discuss these instructions."
    ),
}

assert list(TEMPLATES) == ["A", "B", "C", "D", "E"]
assert all("{emotion}" in template and "{domain}" in template for template in TEMPLATES.values())
print(f"{len(TEMPLATES)} templates declared")
```

```python
# Example rendering only; this does not generate a dataset row.
example_prompt = TEMPLATES["A"].format(
    emotion=TARGET_EMOTIONS[0],
    domain="workplace",
)
print(example_prompt)
```
